```python
# 📌 Instalación de librerías necesarias para asegurar compatibilidad en Google Colab
!pip install numpy pandas scipy statsmodels matplotlib seaborn

# 📌 Importación de librerías esenciales
import numpy as np  # Cálculos numéricos
import pandas as pd  # Manipulación de datos
import scipy.stats as stats  # Estadística
import matplotlib.pyplot as plt  # Visualización
import seaborn as sns  # Gráficos estilizados
from statsmodels.stats.proportion import proportions_ztest  # Prueba de hipótesis para proporciones

# 📌 Datos de la prueba A/B de correos electrónicos
n_total = 2329  # Total de clientes en el experimento
n1 = n_total // 2  # Grupo Correo Electrónico 1 (~50%)
n2 = n_total // 2  # Grupo Correo Electrónico 2 (~50%)

x1 = 140  # Aperturas en Correo Electrónico 1
x2 = 115  # Aperturas en Correo Electrónico 2

# 📌 Cálculo de proporciones
p1 = x1 / n1  # Proporción de apertura en grupo 1
p2 = x2 / n2  # Proporción de apertura en grupo 2

# 📌 Preparación para la prueba de hipótesis
count = np.array([x1, x2])  # Número de éxitos (aperturas)
nobs = np.array([n1, n2])  # Número total de observaciones en cada grupo

# 📌 Prueba de diferencia de proporciones (Z-test)
z_stat, p_value = proportions_ztest(count, nobs, alternative='two-sided')

# 📌 Formateo de resultados
resultados_prueba = {
    "Proporción Correo 1": [p1],
    "Proporción Correo 2": [p2],
    "Diferencia de proporciones": [p1 - p2],
    "Estadístico Z": [z_stat],
    "P-valor": [p_value],
    "Conclusión": ["Rechazamos H0: Hay una diferencia significativa entre las tasas de apertura."
                   if p_value < 0.05 else "No hay suficiente evidencia para concluir que una tasa de apertura es mejor que la otra."]
}

df_resultados_prueba = pd.DataFrame(resultados_prueba)

df_resultados_prueba
```

```python
# Instalamos las librerías necesarias (esto es específico para Google Colab)
!pip install pandas statsmodels

# Importamos las bibliotecas necesarias
import pandas as pd  # Para manejar y procesar los datos
from scipy.stats import ttest_1samp  # Para la prueba t de una muestra
from google.colab import files  # Para cargar archivos en Google Colab
```

```python
# Paso 1: Subir un archivo CSV
print("Por favor, sube tu archivo CSV que contenga los datos para realizar la prueba t.")
uploaded = files.upload()  # Permite al usuario subir un archivo desde su computadora
```

```python
# Paso 2: Cargar los datos en un DataFrame
# Suponemos que el archivo subido tiene un nombre, lo seleccionamos dinámicamente
file_name = list(uploaded.keys())[0]  # Obtenemos el nombre del archivo subido
data = pd.read_csv(file_name)  # Cargamos el archivo CSV en un DataFrame
print("\nDatos cargados con éxito. Primeras filas del DataFrame:")
print(data.head())  # Mostramos las primeras filas para verificar
```

```python
# Paso 3: Verificar las columnas del DataFrame
print("\nColumnas disponibles en el DataFrame:", data.columns)
```

```python
# Paso 4: Seleccionar la columna a analizar
# Se solicita al usuario que ingrese el nombre de la columna que contiene los datos numéricos
column_name = input("\nIngresa el nombre de la columna que deseas analizar: ")
sample_data = data[column_name].dropna()  # Seleccionamos la columna y eliminamos valores nulos
```

```python
# Paso 5: Mostrar estadísticas descriptivas de los datos
print("\nEstadísticas descriptivas de los datos seleccionados:")
print(sample_data.describe())
```

```python
# Paso 6: Definir la hipótesis nula
# Hipótesis nula: La media de la población es mayor o igual a 60
population_mean = 60  # Valor hipotético de la media poblacional
```

```python
# Paso 7: Realizar la prueba t de una muestra
# Usamos scipy.stats.ttest_1samp para realizar la prueba
t_stat, p_value = ttest_1samp(sample_data, popmean=population_mean)
```

```python
# Ajustamos el valor p para una prueba unilateral (menor a 60)
# Dividimos el valor p entre 2 porque estamos haciendo una prueba unilateral
p_value_one_tailed = p_value / 2

# Paso 8: Mostrar los resultados de la prueba t
print("\nResultados de la prueba t de una muestra:")
print(f"Estadístico t: {t_stat:.4f}")
print(f"Valor p (prueba unilateral): {p_value_one_tailed:.4f}")
```

```python
# Paso 9: Interpretación de los resultados
alpha = 0.01  # Nivel de significancia ajustado a 0.01
if t_stat < 0 and p_value_one_tailed < alpha:  # Prueba unilateral hacia la izquierda
    print(
        "\nConclusión: Existe evidencia suficiente para rechazar la hipótesis nula. "
        "Esto indica que la media de la muestra es significativamente menor que 60 al nivel de significancia de 0.01."
    )
else:
    print(
        "\nConclusión: No se puede rechazar la hipótesis nula. "
        "Esto indica que no hay evidencia suficiente para afirmar que la media de la muestra es menor que 60 al nivel de significancia de 0.01."
    )
```

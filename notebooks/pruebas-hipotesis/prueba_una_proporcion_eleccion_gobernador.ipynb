```python
# Script interactivo para prueba de hipótesis para proporciones utilizando statsmodels
# Este script permite al usuario proporcionar valores directamente desde la interfaz en Google Colab.

# Instalación de la librería statsmodels
!pip install statsmodels

# Importar las bibliotecas necesarias
import statsmodels.api as sm

# Solicitar datos del usuario
print("Ingrese los valores necesarios para la prueba de hipótesis:")

# Proporción hipotética bajo H0 (valor teórico que queremos contrastar)
p_0 = float(input("Ingrese la proporción hipotética (p_0, entre 0 y 1): "))

# Proporción observada en la muestra (resultado del sondeo o experimento)
p_muestra = float(input("Ingrese la proporción observada (p_muestra, entre 0 y 1): "))

# Tamaño de la muestra (número total de observaciones)
n = int(input("Ingrese el tamaño de la muestra (n): "))

# Nivel de significancia (probabilidad de cometer un error tipo I, por defecto 0.05)
try:
    alpha = float(input("Ingrese el nivel de significancia (alpha, entre 0 y 1, por defecto 0.05): "))
except ValueError:
    alpha = 0.05  # Si el usuario no ingresa un valor válido, se usa el valor por defecto

# Validar los datos ingresados
# Verificamos que la proporción hipotética esté entre 0 y 1
if not (0 <= p_0 <= 1):
    raise ValueError("La proporción hipotética p_0 debe estar entre 0 y 1.")
# Verificamos que la proporción observada esté entre 0 y 1
if not (0 <= p_muestra <= 1):
    raise ValueError("La proporción observada p_muestra debe estar entre 0 y 1.")
# Verificamos que el tamaño de la muestra sea mayor que 0
if n <= 0:
    raise ValueError("El tamaño de la muestra n debe ser mayor a 0.")

# Calcular el número de éxitos en la muestra (x_muestra)
# Esto se obtiene multiplicando la proporción observada por el tamaño de la muestra
x_muestra = int(p_muestra * n)

# Tamaño total de la muestra
n_muestra = n  # Este valor ya lo ingresó el usuario

# Realizar la prueba de hipótesis usando statsmodels
# proportions_ztest realiza pruebas z para proporciones, útil para este caso
prueba = sm.stats.proportions_ztest(
    count=x_muestra,    # Número de éxitos observados en la muestra
    nobs=n_muestra,     # Tamaño total de la muestra
    value=p_0,          # Proporción hipotética bajo H0
    alternative='smaller'  # Prueba unilateral: evaluamos si la proporción es menor que p_0
)

# Extraer el estadístico z y el p-valor del resultado de la prueba
z_stat_sm, p_val_sm = prueba

# Decisión basada en el p-valor y el nivel de significancia (alpha)
# Si el p-valor es menor que alpha, rechazamos H0 (hipótesis nula)
decision_sm = "Rechazar H0" if p_val_sm < alpha else "No rechazar H0"

# Mostrar los resultados
print("\nResultados de la prueba de hipótesis:")
print(f"Proporción hipotética (p_0): {p_0}")  # Mostrar el valor de p_0
print(f"Proporción observada (p_muestra): {p_muestra}")  # Mostrar el valor de p_muestra
print(f"Tamaño de la muestra (n): {n}")  # Mostrar el tamaño total de la muestra
print(f"Estadístico z: {z_stat_sm:.4f}")  # Mostrar el valor del estadístico z
print(f"P-valor: {p_val_sm:.4f}")  # Mostrar el p-valor asociado al estadístico z
print(f"Decisión: {decision_sm}")  # Mostrar la conclusión sobre H0
```

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

## Cargar dataset (SIEMPRE por subida manual)

```python
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded.keys()))
df = pd.read_csv(file_name)

display(df.head())
df.info()
```

## Preparación mínima

```python
# Validación básica de columnas
required = {"Brand", "Distance", "Durability"}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Faltan columnas: {missing}. Se requieren: {required}")

# Asegurar tipos correctos
df["Brand"] = df["Brand"].astype("category")
df["Distance"] = pd.to_numeric(df["Distance"], errors="coerce")
df["Durability"] = pd.to_numeric(df["Durability"], errors="coerce")

# Eliminar filas con nulos en columnas clave
df = df.dropna(subset=["Brand", "Distance", "Durability"]).copy()

print(f"Filas válidas tras limpieza: {len(df)}")
```

## ANOVA 1: Distance ~ Brand (una vía)

```python
alpha = 0.05

model_dist = smf.ols("Distance ~ C(Brand)", data=df).fit()
anova_dist = sm.stats.anova_lm(model_dist, typ=2)

print("ANOVA: Distance ~ Brand")
display(anova_dist)

p_dist = float(anova_dist.loc["C(Brand)", "PR(>F)"])
print(f"p-value = {p_dist:.6g}")

if p_dist < alpha:
    print(f"Conclusión (α={alpha}): Se rechaza H0. Hay diferencias entre marcas en Distance.")
else:
    print(f"Conclusión (α={alpha}): No se rechaza H0. No hay evidencia de diferencias entre marcas en Distance.")
```

## ANOVA 2: Durability ~ Brand (una vía)

```python
model_dur = smf.ols("Durability ~ C(Brand)", data=df).fit()
anova_dur = sm.stats.anova_lm(model_dur, typ=2)

print("ANOVA: Durability ~ Brand")
display(anova_dur)

p_dur = float(anova_dur.loc["C(Brand)", "PR(>F)"])
print(f"p-value = {p_dur:.6g}")

if p_dur < alpha:
    print(f"Conclusión (α={alpha}): Se rechaza H0. Hay diferencias entre marcas en Durability.")
else:
    print(f"Conclusión (α={alpha}): No se rechaza H0. No hay evidencia de diferencias entre marcas en Durability.")
```

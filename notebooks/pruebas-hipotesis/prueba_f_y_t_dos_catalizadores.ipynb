```python
# Script para evaluar varianzas y medias usando statsmodels en Google Colab
# Este script realiza primero una prueba de hipótesis para comparar dos varianzas (prueba F)
# y luego una prueba t para comparar las medias de dos muestras independientes.

# Instalación de las librerías necesarias
!pip install statsmodels pandas numpy

# Importar las bibliotecas necesarias
import numpy as np
import pandas as pd
from scipy.stats import f_oneway, ttest_ind
from statsmodels.stats.weightstats import CompareMeans, DescrStatsW
from google.colab import files

# Función para cargar datos desde un archivo CSV
def cargar_datos():
    """
    Permite al usuario subir un archivo CSV y cargar los datos en un DataFrame.

    Retorna:
    - datos: DataFrame que contiene los datos cargados desde el archivo CSV.
    """
    print("Suba un archivo CSV que contenga las columnas de los datos de las dos muestras.")
    archivo = files.upload()  # Subir archivo desde la computadora
    nombre_archivo = list(archivo.keys())[0]  # Obtener el nombre del archivo subido
    # Leer el archivo en un DataFrame
    datos = pd.read_csv(nombre_archivo)
    print("Archivo cargado exitosamente.")
    print(datos.head())  # Mostrar las primeras filas para verificar el contenido
    return datos

# Función para realizar la prueba F para comparar varianzas
def prueba_varianzas(datos, columna1, columna2, alpha=0.05):
    """
    Realiza una prueba F para evaluar si hay una diferencia significativa entre las varianzas de dos muestras.

    Parámetros:
    - datos: DataFrame con los datos.
    - columna1: Nombre de la columna para la primera muestra.
    - columna2: Nombre de la columna para la segunda muestra.
    - alpha: Nivel de significancia (por defecto 0.05).

    Retorna:
    - resultado: Cadena indicando si se rechaza o no la hipótesis nula.
    """
    # Extraer los datos de las columnas especificadas
    muestra1 = datos[columna1].dropna()
    muestra2 = datos[columna2].dropna()

    # Calcular la estadística F
    f_stat = np.var(muestra1, ddof=1) / np.var(muestra2, ddof=1)
    df1 = len(muestra1) - 1  # Grados de libertad para la primera muestra
    df2 = len(muestra2) - 1  # Grados de libertad para la segunda muestra

    # Calcular el valor p usando la distribución F
    p_val = f_oneway(muestra1, muestra2).pvalue

    print("\nResultados de la prueba F para comparar varianzas:")
    print(f"Estadístico F: {f_stat:.4f}")
    print(f"P-valor: {p_val:.4f}")

    # Decisión basada en el p-valor
    if p_val < alpha:
        resultado = "Rechazar H0: Existe una diferencia significativa en las varianzas."
    else:
        resultado = "No rechazar H0: No hay evidencia suficiente para concluir que las varianzas son diferentes."

    print(resultado)
    return resultado

# Función para realizar la prueba t para comparar medias
def prueba_medias(datos, columna1, columna2, alpha=0.05):
    """
    Realiza una prueba t para comparar las medias de dos muestras independientes.

    Parámetros:
    - datos: DataFrame con los datos.
    - columna1: Nombre de la columna para la primera muestra.
    - columna2: Nombre de la columna para la segunda muestra.
    - alpha: Nivel de significancia (por defecto 0.05).

    Retorna:
    - resultado: Cadena indicando si se rechaza o no la hipótesis nula.
    """
    # Extraer los datos de las columnas especificadas
    muestra1 = datos[columna1].dropna()
    muestra2 = datos[columna2].dropna()

    # Calcular estadísticas descriptivas para las dos muestras
    descr1 = DescrStatsW(muestra1)
    descr2 = DescrStatsW(muestra2)

    # Comparar las medias utilizando statsmodels
    cm = CompareMeans(descr1, descr2)
    t_stat, p_val, _ = cm.ttest_ind(alternative='two-sided', usevar='pooled')

    print("\nResultados de la prueba t para comparar medias:")
    print(f"Estadístico t: {t_stat:.4f}")
    print(f"P-valor: {p_val:.4f}")

    # Decisión basada en el p-valor
    if p_val < alpha:
        resultado = "Rechazar H0: Existe una diferencia significativa en las medias."
    else:
        resultado = "No rechazar H0: No hay evidencia suficiente para concluir que las medias son diferentes."

    print(resultado)
    return resultado

# Ejecución directa
# Cargar los datos directamente
datos = cargar_datos()  # Llama a la función para subir y cargar los datos

# Solicitar al usuario los nombres de las columnas para las muestras
columna1 = input("Ingrese el nombre de la columna para la primera muestra: ")
columna2 = input("Ingrese el nombre de la columna para la segunda muestra: ")

# Realizar la prueba de comparación de varianzas
prueba_varianzas(datos, columna1, columna2)

# Realizar la prueba t para comparar medias
prueba_medias(datos, columna1, columna2)
```

```python
# Script para prueba de hipótesis con statsmodels en Google Colab
# Este script permite subir un archivo con los datos y realizar una prueba t de hipótesis
# para determinar si la vida útil media de las baterías excede un valor dado.

# Instalación de las librerías necesarias
!pip install statsmodels pandas numpy

# Importar las bibliotecas necesarias
import numpy as np
import pandas as pd
from statsmodels.stats.weightstats import DescrStatsW
from google.colab import files

# Función para subir archivo de datos
def cargar_datos():
    """
    Permite al usuario subir un archivo CSV y cargar los datos en un DataFrame.

    Retorna:
    - datos: DataFrame que contiene los datos cargados desde el archivo CSV.
    """
    print("Suba un archivo CSV que contenga una columna con los datos de vida útil de las baterías.")
    # Solicitar al usuario subir el archivo
    archivo = files.upload()  # Subir archivo desde la computadora
    nombre_archivo = list(archivo.keys())[0]  # Obtener el nombre del archivo subido
    # Leer el archivo en un DataFrame
    datos = pd.read_csv(nombre_archivo)
    print("Archivo cargado exitosamente.")
    print(datos.head())  # Mostrar las primeras filas para verificar el contenido
    return datos

# Función para realizar la prueba de hipótesis
def prueba_hipotesis(datos, columna, mu_0, alpha=0.05):
    """
    Realiza una prueba t unilateral para determinar si la media poblacional es mayor a un valor dado.

    Parámetros:
    - datos: DataFrame con los datos.
    - columna: Nombre de la columna que contiene las vidas útiles.
    - mu_0: Valor de referencia para la hipótesis nula (H0).
    - alpha: Nivel de significancia (por defecto 0.05).
    """
    # Extraer los datos de la columna especificada, eliminando valores nulos si los hubiera
    muestras = datos[columna].dropna()

    # Crear un objeto de estadísticas descriptivas para los datos
    # Esto permite calcular la media, desviación estándar, y realizar pruebas t
    descr_stats = DescrStatsW(muestras)

    # Realizar la prueba t unilateral
    # H0: La media es menor o igual a mu_0
    # Ha: La media es mayor a mu_0
    t_stat, p_val, _ = descr_stats.ttest_mean(mu_0, alternative='larger')

    # Resultados de la prueba
    print("\nResultados de la prueba de hipótesis:")
    print(f"Media muestral: {descr_stats.mean:.2f}")  # Calcular la media de la muestra
    print(f"Desviación estándar: {descr_stats.std:.2f}")  # Calcular la desviación estándar
    print(f"Estadístico t: {t_stat:.4f}")  # Mostrar el valor del estadístico t
    print(f"P-valor: {p_val:.4f}")  # Mostrar el valor p para la prueba unilateral

    # Decisión basada en el p-valor
    if p_val < alpha:
        print(f"\nCon un nivel de significancia de {alpha}, rechazamos la hipótesis nula (H0).")
        print("Conclusión: Hay evidencia suficiente para decir que la vida útil media excede el valor de referencia.")
    else:
        print(f"\nCon un nivel de significancia de {alpha}, no rechazamos la hipótesis nula (H0).")
        print("Conclusión: No hay evidencia suficiente para decir que la vida útil media excede el valor de referencia.")

# Ejecución directa
# Cargar los datos directamente
datos = cargar_datos()  # Llama a la función para subir y cargar los datos

# Solicitar al usuario el nombre de la columna de interés
columna_datos = input("Ingrese el nombre de la columna con las vidas útiles: ")

# Solicitar al usuario el valor de referencia para la hipótesis nula
mu_0 = float(input("Ingrese el valor de referencia para la hipótesis nula (mu_0): "))

# Realizar la prueba de hipótesis
prueba_hipotesis(datos, columna_datos, mu_0)
```

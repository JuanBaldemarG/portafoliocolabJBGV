```python
# 📌 Instalamos la librería 'statsmodels' en caso de que no esté instalada
# Esto es necesario para realizar la prueba de hipótesis de dos proporciones en Google Colab
!pip install statsmodels

# 📌 Importamos las librerías necesarias
import numpy as np  # Para cálculos numéricos
import statsmodels.api as sm  # Para realizar la prueba estadística
```

```python
# 📌 Definimos los datos del experimento
# k1 y n1 corresponden al proceso 1 (cantidad de defectos y total de productos inspeccionados)
# k2 y n2 corresponden al proceso 2 (cantidad de defectos y total de productos inspeccionados)
k1, n1 = 120, 400  # Datos del Proceso 1 (antes del cambio)
k2, n2 = 80, 400   # Datos del Proceso 2 (después del cambio)
```

```python
# 📌 Calculamos las proporciones de defectos en cada proceso
p1 = k1 / n1  # Proporción de defectos en el Proceso 1
p2 = k2 / n2  # Proporción de defectos en el Proceso 2
```

```python
# 📌 Creamos los arreglos para la prueba estadística
count = np.array([k1, k2])  # Cantidad de defectos en cada proceso
nobs = np.array([n1, n2])   # Número total de productos inspeccionados en cada proceso
```

```python
# 📌 Realizamos la prueba de hipótesis de dos proporciones
# Usamos la prueba Z para comparar dos proporciones con la función 'proportions_ztest'
# El argumento 'alternative="larger"' indica que estamos probando si p1 > p2 (prueba de una cola)
z_stat, p_value = sm.stats.proportions_ztest(count, nobs, alternative='larger')
```

```python
# 📌 Definimos el nivel de significancia (α)
alpha = 0.05  # 5% de significancia (confianza del 95%)
```

```python
# 📌 Mostramos los resultados
print("📊 RESULTADOS DE LA PRUEBA DE HIPÓTESIS 📊")
print(f"🔹 Proporción de defectos - Proceso 1 (antes del cambio): {p1:.3f}")
print(f"🔹 Proporción de defectos - Proceso 2 (después del cambio): {p2:.3f}")
print(f"🔹 Estadístico Z: {z_stat:.3f}")
print(f"🔹 P-valor: {p_value:.5f}")
```

```python
# 📌 Interpretación de los resultados
print("\n📌 CONCLUSIÓN:")
if p_value < alpha:
    print("✅ Rechazamos H0: El Proceso 1 tiene una proporción de defectos significativamente mayor que el Proceso 2.")
else:
    print("❌ No hay suficiente evidencia para concluir que el Proceso 1 tenga más defectos que el Proceso 2.")
```

# Caso práctico RFM en Google Colab (CSV o XLSX)

## Contexto del caso
Imagina que trabajas en el equipo de analítica de una tienda e-commerce.
En los últimos meses, el costo de adquisición subió y el presupuesto de marketing es limitado.
La empresa necesita decidir **a qué clientes priorizar** para maximizar ingresos con campañas más eficientes.

### Problema de negocio
No todos los clientes tienen el mismo valor ni el mismo riesgo de abandono.
Si tratamos a todos igual, desperdiciamos presupuesto y reducimos el ROI de marketing.

### Objetivo analítico
Segmentar clientes con RFM para responder:
- ¿Quién compra recientemente y con alta frecuencia?
- ¿Quién gastó mucho pero lleva tiempo sin comprar?
- ¿Qué clientes requieren reactivación inmediata?

### Definiciones RFM
- R (Recency): días desde la última compra (menor es mejor).
- F (Frequency): número de compras o facturas (mayor es mejor).
- M (Monetary): gasto total acumulado (mayor es mejor).


## Instrucciones de uso
1. Ejecuta la celda de carga y sube un archivo .csv o .xlsx.
2. Verifica que tenga columnas compatibles: InvoiceNo, CustomerID, InvoiceDate, Quantity, UnitPrice.
3. Ejecuta todas las celdas en orden para obtener segmentos, interpretación y conclusiones.


```python
import pandas as pd
from google.colab import files

print('Sube un archivo .csv o .xlsx')
uploaded = files.upload()
filename = next(iter(uploaded))

if filename.lower().endswith('.csv'):
    df = pd.read_csv(filename)
elif filename.lower().endswith('.xlsx'):
    df = pd.read_excel(filename)
else:
    raise ValueError('Formato no soportado. Usa .csv o .xlsx')

print(f'Archivo cargado: {filename}')
print('Dimensión original:', df.shape)
df.head()
```

### Explicación de la celda de carga
- files.upload abre el selector para subir archivos locales a Colab.
- El script detecta automáticamente si el archivo es CSV o XLSX.
- Se imprime la dimensión inicial para tener una línea base antes de limpiar.


## Validación mínima de estructura
Esta validación evita errores aguas abajo y mejora la trazabilidad del análisis.


```python
required_cols = ['InvoiceNo', 'CustomerID', 'InvoiceDate', 'Quantity', 'UnitPrice']
missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f'Faltan columnas requeridas: {missing}')
print('Columnas requeridas presentes.')
```

## Limpieza para RFM
Objetivo: conservar transacciones válidas que representen compras reales.

Reglas aplicadas:
1. Convertir InvoiceDate a fecha.
2. Eliminar nulos en CustomerID e InvoiceDate.
3. Excluir cancelaciones (InvoiceNo inicia con C).
4. Excluir Quantity y UnitPrice no positivos.
5. Crear MonetaryValue = Quantity por UnitPrice.


```python
before_rows = len(df)

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
df = df.dropna(subset=['CustomerID', 'InvoiceDate'])
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.startswith('C', na=False)]
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
df['MonetaryValue'] = df['Quantity'] * df['UnitPrice']

after_rows = len(df)
print('Filas antes de limpiar:', before_rows)
print('Filas después de limpiar:', after_rows)
print('Filas removidas:', before_rows - after_rows)
df.head()
```

### Interpretación de limpieza
- Si se removieron demasiadas filas, revisa calidad del origen (errores de captura o estructura).
- Si casi no se removieron filas, el dato está relativamente sano para segmentación inicial.
- La limpieza impacta directamente la calidad del score y las decisiones de negocio.


## Construcción de métricas RFM
- Recency: qué tan reciente fue la última compra.
- Frequency: cuántas veces compró cada cliente.
- Monetary: cuánto dinero dejó cada cliente.


```python
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)
rfm = df.groupby('CustomerID').agg(
    Recency=('InvoiceDate', lambda x: (snapshot_date - x.max()).days),
    Frequency=('InvoiceNo', 'nunique'),
    Monetary=('MonetaryValue', 'sum')
).reset_index()

print('Clientes únicos:', rfm['CustomerID'].nunique())
rfm.head()
```

## Scoring RFM (1 a 5)
Usamos quintiles para normalizar y comparar clientes en una escala común.

Criterio:
- R: menor recencia = mejor score (5).
- F y M: mayor valor = mejor score (5).


```python
rfm['R_Score'] = pd.qcut(rfm['Recency'].rank(method='first'), 5, labels=[5,4,3,2,1]).astype(int)
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5]).astype(int)
rfm['M_Score'] = pd.qcut(rfm['Monetary'].rank(method='first'), 5, labels=[1,2,3,4,5]).astype(int)
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)
rfm.sort_values(['RFM_Score', 'Monetary'], ascending=[False, False]).head(10)
```

## Segmentación de negocio
Reglas base de este ejercicio (puedes ajustarlas por industria):
- Champions: R>=4, F>=4, M>=4
- Loyal Customers: R>=3 y F>=3
- At Risk: R<=2 y F>=3
- Hibernating: resto


```python
def segment(row):
    if row['R_Score'] >= 4 and row['F_Score'] >= 4 and row['M_Score'] >= 4:
        return 'Champions'
    elif row['R_Score'] >= 3 and row['F_Score'] >= 3:
        return 'Loyal Customers'
    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'
    else:
        return 'Hibernating'

rfm['Segment'] = rfm.apply(segment, axis=1)

summary = rfm.groupby('Segment').agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).sort_values('Customers', ascending=False)

summary
```

## Interpretación de resultados (guía)
Usa esta guía al leer la tabla summary:

- Si Champions tiene alto Avg_Monetary y baja Avg_Recency: cartera premium activa.
- Si At Risk tiene buen Avg_Monetary pero alta Avg_Recency: hay riesgo real de pérdida de valor.
- Si Hibernating concentra muchos clientes: oportunidad de campañas automatizadas de reactivación.
- Si Loyal Customers es grande: base sólida para cross-sell y bundles.


```python
top_customers = rfm.sort_values('Monetary', ascending=False).head(20)
segment_share = (summary['Customers'] / summary['Customers'].sum() * 100).round(2)

print('Participación por segmento (%)')
display(segment_share.to_frame('Share_%'))

print('Top 20 clientes por gasto:')
top_customers
```

## Recomendaciones accionables por segmento
- Champions: programas VIP, acceso anticipado, upselling premium.
- Loyal Customers: cross-sell, beneficios por recurrencia, referidos.
- At Risk: ofertas de retorno con urgencia temporal y comunicación personalizada.
- Hibernating: campañas de bajo costo, reactivación gradual y limpieza de base.

### KPI sugeridos para seguimiento
- Tasa de reactivación a 30/60/90 días.
- Ingreso incremental por segmento.
- Retención mensual de Champions y Loyal.
- Costo por recuperación de At Risk.


```python
rfm.to_csv('rfm_resultados.csv', index=False)
summary.to_csv('rfm_resumen_segmentos.csv')
print('Guardado: rfm_resultados.csv y rfm_resumen_segmentos.csv')
```

## Conclusiones generales ampliadas
1. RFM traduce datos transaccionales en decisiones de negocio concretas y priorizadas.
2. El valor real del enfoque no es solo clasificar clientes, sino **asignar presupuesto con mayor eficiencia**.
3. Recency suele anticipar comportamiento de corto plazo, mientras Frequency y Monetary explican valor acumulado.
4. Un cliente de alto Monetary con recencia alta es una alerta estratégica: puede requerir intervención inmediata.
5. RFM es una excelente base para madurez analítica: CLV, modelos de churn, propensión de compra y personalización.
6. La calidad del dato (limpieza y consistencia) condiciona directamente la calidad de la segmentación.
7. Este framework debe revisarse periódicamente (mensual o trimestral) para capturar cambios en el comportamiento de compra.
